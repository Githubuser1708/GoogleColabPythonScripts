```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.getOrCreate()
```

```python
spark
```

```python
#read sklearn inbuilt data
import pandas as pd
import numpy as np
from sklearn.datasets import load_wine
```

```python
wine = load_wine(as_frame=True)
wine = wine.frame
wine = spark.createDataFrame(wine)
```

```python
wine.show()
```

```python
wine.printSchema()
```

```python
from pyspark.ml.feature import VectorAssembler
```

```python
wine.columns
```

```python
featureassembler = VectorAssembler(inputCols=['alcohol',
 'malic_acid',
 'ash',
 'alcalinity_of_ash',
 'magnesium',
 'total_phenols',
 'flavanoids',
 'nonflavanoid_phenols',
 'proanthocyanins',
 'color_intensity',
 'hue',
 'od280/od315_of_diluted_wines',
 'proline',
 'target'], outputCol="Features")
```

```python
output = featureassembler.transform(wine)
```

```python
output.show()
```

```python
modeldata=output.select("Features","target")
```

```python
modeldata.show()
```

```python
train_data,test_data=modeldata.randomSplit([0.8,0.2])
```

```python
train_data.show()
```

```python
test_data.show()
```

```python
from pyspark.ml.classification import RandomForestClassifier
```

```python
rf = RandomForestClassifier(featuresCol="Features",labelCol="target" )
```

```python
rf=rf.fit(train_data)
```

```python
y_pred = rf.transform(test_data)
```

```python
y_pred.show()
```

```python
y_pred.groupBy("target","prediction").count().show()
```

```python
from sklearn.metrics import confusion_matrix
pred=y_pred.select("prediction").collect()
orig=y_pred.select("target").collect()
print(confusion_matrix(orig,pred))
```

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
```

```python
evaluator = MulticlassClassificationEvaluator(labelCol="target", predictionCol="prediction" )
```

```python
accuracy = evaluator.evaluate(y_pred)
```

```python
accuracy
```

```python
spark.stop()
```
